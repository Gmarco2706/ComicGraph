#PULIZIA DEI DATI DEL DATASET FUMETTI MARVEL

```python
from pathlib import Path

import pandas as pd
from pathlib import Path

import pandas as pd

BASE_PATH = Path('/Users/marcogreco/ComicGraph')
MARVEL_COMICS_PATH = BASE_PATH / 'Dataset' / 'Dataset_Originali' / 'Marvel_Comics.csv'

MARVEL_COMICS = pd.read_csv(MARVEL_COMICS_PATH)

print('Numero di righe:', len(MARVEL_COMICS))
print('Numero di colonne:', len(MARVEL_COMICS.columns))
print('Colonne:', MARVEL_COMICS.columns.tolist())

```

```python
# Rimozione dei duplicati completi
rows_before = len(MARVEL_COMICS)
duplicates_removed = int(MARVEL_COMICS.duplicated().sum())

MARVEL_COMICS = MARVEL_COMICS.drop_duplicates().reset_index(drop=True)

print('Duplicati rimossi:', duplicates_removed)
print('Righe prima della rimozione:', rows_before)
print('Righe dopo la rimozione:', len(MARVEL_COMICS))
print('Duplicati residui:', int(MARVEL_COMICS.duplicated().sum()))
```

```python
#rimozione colonne rating e imprint per valori mancanti e perchè imprint era un dato molto sporco e non presente sull'analogo dataset relativo alla DC
MARVEL_COMICS = MARVEL_COMICS.drop(columns=['Rating'])
MARVEL_COMICS = MARVEL_COMICS.drop(columns=['Imprint'])
MARVEL_COMICS = MARVEL_COMICS.drop(columns=['Price']
)
print('Dimensioni dopo la rimozione di Rating:', MARVEL_COMICS.shape)
print('Colonne attuali:', MARVEL_COMICS.columns.tolist())
print('Dimensioni dopo la rimozione di imprint:', MARVEL_COMICS.shape)
print('Colonne attuali:', MARVEL_COMICS.columns.tolist())
print('Dimensioni dopo la rimozione di price:', MARVEL_COMICS.shape)
print('Colonne attuali:', MARVEL_COMICS.columns.tolist())
```

```python
# Gestione dei valori mancanti: riempimento con 'Unknown'
columns_to_fill_marvel = [
    'cover_artist',
    'penciler',
    'writer',
    'issue_description',
    'Format',
    'issue_title',
    'publish_date',
    'comic_name',
    'active_years'
]

MARVEL_COMICS[columns_to_fill_marvel] = MARVEL_COMICS[columns_to_fill_marvel].fillna('Unknown')

print("Valori mancanti dopo il fill in queste colonne:")
print(MARVEL_COMICS[columns_to_fill_marvel].isna().sum())
```

```python
# Salvataggio del dataset Marvel Comics pulito
OUTPUT_PATH = BASE_PATH / 'Dataset' / 'Puliti'
OUTPUT_PATH.mkdir(exist_ok=True)

MARVEL_CLEAN_PATH = OUTPUT_PATH / 'Marvel_Comics_clean.csv'
MARVEL_COMICS.to_csv(MARVEL_CLEAN_PATH, index=False)

print(f'Dataset Marvel Comics salvato in: {MARVEL_CLEAN_PATH}')
print(f'Righe finali: {len(MARVEL_COMICS)}')
print(f'Colonne: {MARVEL_COMICS.columns.tolist()}')
```
