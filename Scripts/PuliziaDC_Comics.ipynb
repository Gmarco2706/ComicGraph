# Pulizia Dc_comics.csv

```python
from pathlib import Path

import pandas as pd

BASE_PATH = Path('/Users/marcogreco/ComicGraph')
DC_COMICS_PATH = BASE_PATH / 'Dataset' / 'Dataset_Originali' / 'DC_Comics.csv'

DC_COMICS = pd.read_csv(DC_COMICS_PATH)

DC_COMICS = DC_COMICS.drop(columns=['Rating'])
#i valori nulli sono stati modificati in valori "unknown"
columns_to_fill = [
    'Colourists',
    'Letterers',
    'Executive_Editor',
    'Editors',
    'Writers',
    'Inkers',
    'Cover_Artists',
    'Pencilers',
    'Release_Date',
]

DC_COMICS[columns_to_fill] = DC_COMICS[columns_to_fill].fillna('Unknown')

print('Dimensioni dopo la pulizia:', DC_COMICS.shape)
print('Colonne:', DC_COMICS.columns.tolist())
```

```python
# Rimozione dei duplicati completi
rows_before = len(DC_COMICS)
duplicates_removed = int(DC_COMICS.duplicated().sum())

DC_COMICS = DC_COMICS.drop_duplicates().reset_index(drop=True)

print('Duplicati rimossi:', duplicates_removed)
print('Righe prima della rimozione:', rows_before)
print('Righe dopo la rimozione:', len(DC_COMICS))
print('Duplicati residui:', int(DC_COMICS.duplicated().sum()))
```

```python
# Rimozione delle righe senza informazioni sull'issue
unavailable_issue_mask = (
    DC_COMICS['Issue_Name'].eq('Not Available')
    & DC_COMICS['Issue_Link'].eq('Not Available')
)

unavailable_issue_rows = int(unavailable_issue_mask.sum())
DC_COMICS = DC_COMICS.loc[~unavailable_issue_mask].reset_index(drop=True)

print('Righe Not Available rimosse:', unavailable_issue_rows)
print('Righe dopo la rimozione:', len(DC_COMICS))
print('Righe Not Available residue:', int(
    (
        DC_COMICS['Issue_Name'].eq('Not Available')
        & DC_COMICS['Issue_Link'].eq('Not Available')
    ).sum()
))
```

```python
# Salvataggio del dataset DC Comics pulito
OUTPUT_PATH = BASE_PATH / 'Dataset' / 'Puliti'
OUTPUT_PATH.mkdir(exist_ok=True)

DC_CLEAN_PATH = OUTPUT_PATH / 'DC_Comics_clean.csv'
DC_COMICS.to_csv(DC_CLEAN_PATH, index=False)

print(f'Dataset DC Comics salvato in: {DC_CLEAN_PATH}')
print(f'Righe finali: {len(DC_COMICS)}')
print(f'Colonne: {DC_COMICS.columns.tolist()}')
```
