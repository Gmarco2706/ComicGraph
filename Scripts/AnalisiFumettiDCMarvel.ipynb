#Analisi del dataset fumetti DC

```python
from pathlib import Path

import pandas as pd

BASE_PATH = Path('/Users/marcogreco/ComicGraph')
DATASET_PATH = BASE_PATH / 'Dataset' / 'Dataset_Originali'

MARVEL_COMICS_PATH = DATASET_PATH / 'Marvel_Comics.csv'
DC_COMICS_PATH = DATASET_PATH / 'DC_Comics.csv'
```

```python
# Caricamento del dataset DC Comics
DC_COMICS = pd.read_csv(DC_COMICS_PATH)

print('Numero di righe:', len(DC_COMICS))
print('Numero di colonne:', len(DC_COMICS.columns))
print('Colonne del dataset:')
print(DC_COMICS.columns.tolist())
```

```python
# Analisi dei valori mancanti per colonna
missing_dc = (
    DC_COMICS.isna()
    .sum()
    .rename('valori_mancanti')
    .to_frame()
)

missing_dc['percentuale'] = (
    missing_dc['valori_mancanti'] / len(DC_COMICS) * 100
).round(2)

missing_dc.sort_values(
    'valori_mancanti',
    ascending=False
)
```

```python
# Analisi dei duplicati completi nel dataset DC
DC_DUPLICATES = DC_COMICS[DC_COMICS.duplicated(keep=False)].copy()

print('Duplicati completi DC:', DC_COMICS.duplicated().sum())
print('Righe coinvolte nei duplicati DC:', len(DC_DUPLICATES))

DC_DUPLICATES.head()
```

```python
# Caricamento e analisi dei duplicati completi nel dataset Marvel
MARVEL_COMICS = pd.read_csv(MARVEL_COMICS_PATH)
MARVEL_DUPLICATES = MARVEL_COMICS[MARVEL_COMICS.duplicated(keep=False)].copy()

print('Duplicati completi Marvel:', MARVEL_COMICS.duplicated().sum())
print('Righe coinvolte nei duplicati Marvel:', len(MARVEL_DUPLICATES))

MARVEL_DUPLICATES.head()
```

```python
# Analisi delle stringhe speciali Not Available
special_value = 'Not Available'

special_values_report = []
for brand, dataframe in [('DC', DC_COMICS), ('Marvel', MARVEL_COMICS)]:
    matches_by_column = (dataframe == special_value).sum()
    for column, count in matches_by_column[matches_by_column > 0].items():
        special_values_report.append({
            'dataset': brand,
            'colonna': column,
            'valore': special_value,
            'occorrenze': int(count),
        })

special_values_report = pd.DataFrame(special_values_report)

if special_values_report.empty:
    print('La stringa Not Available non è presente nei due dataset.')
else:
    print(special_values_report.to_string(index=False))
```
